```python
import numpy as np # for doing stuff with numbers
import matplotlib.pyplot as plt # for plotting
from ipywidgets import interact, IntSlider

# load the saved data
# we like this npz format better because it allows us to inherit the metadata of the simulation...
data = np.load('qw_history.npz')
probabilities_history = data['probs']
times = data['times']
num_sites = data['num_sites']
boundary_width = data['boundary_width']

positions = np.arange(num_sites) - int(num_sites/2)

# this is the recipe to be used at every fixed position of the slider
def plot_frame(step):
    plt.figure(figsize=(10, 6))

    # probability at the current time step
    current_probs = probabilities_history[step]
    current_t = times[step]

    plt.bar(positions, current_probs, color='red', edgecolor='black', alpha=0.7)
    plt.title(f"CTQW on an $N={num_sites}$ ring (at $t={current_t:.1f}$)")
    plt.xlabel("$j$")
    plt.axvline(x=boundary_width-num_sites/2)
    plt.axvline(x=num_sites-boundary_width-num_sites/2)
    plt.ylabel(rf"$p_j = |\psi_j|^2$")

    start_tick = int(np.floor(positions[0] / 5) * 5)
    end_tick = int(np.ceil(positions[-1] / 5) * 5)
    ticks = np.arange(start_tick, end_tick + 1, 5)
    labels = [str(t) if t % 10 == 0 else '' for t in ticks]
    plt.xticks(ticks, labels)

    # fixing the y-axis so that it does not rescale when using the slider
    plt.ylim(0, 1.0)
    plt.grid(axis='y', linestyle='--', alpha=0.5)
    plt.show()

# interactive slider to scroll time
interact(plot_frame, step=IntSlider(min=0, max=len(times)-1, step=1, value=0, description='Time Step:'))
```
